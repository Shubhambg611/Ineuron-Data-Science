```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential  # there are numltiple ways of for the solving the problem sequntial staking one layer on each other 
from tensorflow.keras.layers import Dense,Flatten #dense = hidden layer 
# flatten = 2d to 1d data or 3d data to 1d data 
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
```

```python
(X_train,y_train),(X_test,y_test) = keras.datasets.mnist.load_data()
```

```python
# normalization / standardization   values will be come between 0-1
X_train = X_train/255
X_test = X_test/255
```

```python
X_train[1]
```

```python
model = Sequential()

model.add(Flatten(input_shape=(28,28)))   # 2D to 1D 
model.add(Dense(128,activation='relu'))   #  
model.add(Dense(32,activation='relu'))
model.add(Dense(10,activation='softmax'))  # 10 becouse we have 10 Diffrent class 
# becouse of multiclass classification using softmax
```

```python
model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy',optimizer='SGD',metrics=['accuracy'])
# SGD stocastic gradient desent
# loss = 2 types write in book refer it
```

```python
history = model.fit(X_train,y_train,epochs=25,validation_split=0.2)
```

```python
y_prob = model.predict(X_test)
```

```python
y_pred = y_prob.argmax(axis=1)
```

```python
accuracy_score(y_test,y_pred)
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
```
